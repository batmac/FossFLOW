# Introduction didactique à SymPy

Bienvenue dans ce cours interactif sur **SymPy** ! Nous allons explorer cette bibliothèque pas à pas.

### Étape 1 : Importer SymPy et définir des variables symboliques
En Python standard, `x` est une variable qui doit contenir une valeur numérique ou un objet. Avec SymPy, nous définissons des objets de type `Symbol` pour représenter des variables mathématiques indéterminées.

```python
import sympy as sp

# Initialisation d'un affichage mathématique soigné (LaTeX/Unicode dans Colab)
sp.init_printing(use_unicode=True)

# Définissons notre premier symbole : x
x = sp.Symbol('x')

# Nous pouvons aussi définir plusieurs symboles à la fois avec la fonction symbols
y, z = sp.symbols('y z')

print("Symboles créés avec succès !")
display(x, y, z)
```

### Étape 2 : Créer et évaluer des expressions

Maintenant que nous avons nos symboles, nous pouvons construire des expressions algébriques naturellement.

```python
# Création d'une expression quadratique
expr = x**2 + 2*x + 1

print("Notre expression symbolique :")
display(expr)

# Nous pouvons évaluer cette expression en remplaçant 'x' par une valeur numérique avec .subs()
valeur_en_3 = expr.subs(x, 3)
print("Évaluation pour x = 3 :")
display(valeur_en_3)
```

### Étape 3 : Développement et Factorisation

SymPy excelle dans la manipulation algébrique. Voyons comment développer une expression ou la factoriser.

```python
# Factorisons notre expression précédente (x^2 + 2x + 1)
expr_factorisee = sp.factor(expr)
print("Expression factorisée :")
display(expr_factorisee)

# Développons une autre expression
expr_produit = (x - 1) * (x + 2) * (x - 3)
print("\nProduit original :")
display(expr_produit)

expr_developpee = sp.expand(expr_produit)
print("Expression développée :")
display(expr_developpee)
```

### Étape 4 : Résolution d'équations (Algèbre)

Avec SymPy, nous pouvons résoudre des équations symboliques ou numériques en utilisant la fonction `solveset` ou `solve`. Par défaut, SymPy considère qu'une expression passée en argument est égale à zéro ($f(x) = 0$).

```python
# Résoudre x^2 - 4 = 0
print("Résolution de x^2 - 4 = 0 :")
solution = sp.solveset(x**2 - 4, x)
display(solution)

# Pour poser explicitement une équation du type f(x) = g(x), on utilise Eq
# Exemple : x + 2 = 5
equation = sp.Eq(x + 2, 5)
print("\nRésolution de l'équation x + 2 = 5 :")
solution_eq = sp.solveset(equation, x)
display(solution_eq)
```

### Étape 5 : Analyse (Dérivées et Intégrales)

Le calcul infinitésimal est extrêmement simple et élégant avec SymPy. On utilise `diff` pour dériver et `integrate` pour intégrer.

```python
# Définissons une fonction, par exemple : sin(x) * exp(x)
f = sp.sin(x) * sp.exp(x)
print("Fonction originale f(x) :")
display(f)

# Calcul de la dérivée première par rapport à x
derivee = sp.diff(f, x)
print("\nDérivée f'(x) :")
display(derivee)

# Calcul d'une intégrale indéfinie (primitive)
primitive = sp.integrate(f, x)
print("\nPrimitive de f(x) :")
display(primitive)

# Calcul d'une intégrale définie de 0 à pi
integrale_definie = sp.integrate(sp.sin(x), (x, 0, sp.pi))
print("\nIntégrale de sin(x) de 0 à pi :")
display(integrale_definie)
```

### Étape 6 : Limites et Séries de Taylor

SymPy permet de calculer des limites complexes (y compris aux infinis) et d'approcher des fonctions à l'aide de développements limités (séries de Taylor) grâce à `sp.limit` et `sp.series`.

```python
# 1. Calcul de limite : lim_{x -> 0} sin(x)/x
limite_sin = sp.limit(sp.sin(x)/x, x, 0)
print("Limite de sin(x)/x quand x tend vers 0 :")
display(limite_sin)

# 2. Limite à l'infini : lim_{x -> oo} (1 + 1/x)^x
limite_inf = sp.limit((1 + 1/x)**x, x, sp.oo)
print("\nLimite de (1 + 1/x)^x quand x tend vers l'infini :")
display(limite_inf)

# 3. Développement limité (série de Taylor) de cos(x) autour de 0 à l'ordre 6
serie_cos = sp.series(sp.cos(x), x, 0, 6)
print("\nDéveloppement limité de cos(x) à l'ordre 6 :")
display(serie_cos)
```

### Étape 7 : Algèbre Linéaire (Matrices et Vecteurs)

En calcul symbolique, les matrices peuvent contenir des symboles (variables) et non uniquement des nombres. On utilise l'objet `sp.Matrix`.

```python
# Définition d'une matrice 2x2 contenant des variables
M = sp.Matrix([[x, y], [1, z]])
print("Matrice symbolique M :")
display(M)

# Calcul du déterminant
det_M = M.det()
print("\nDéterminant de M :")
display(det_M)

# Matrice numérique pour les valeurs propres
A = sp.Matrix([[3, -2], [1, 0]])
print("\nMatrice A (numérique) :")
display(A)

# Calcul des valeurs propres (eigenvalues)
valeurs_propres = A.eigenvals()
print("\nValeurs propres de A (avec leur multiplicité) :")
display(valeurs_propres)
```

### Étape 8 : Résolution d'équations différentielles (EDO)

Pour résoudre une équation différentielle, nous devons définir une fonction inconnue (par exemple $y(t)$) et spécifier l'équation à l'aide de sa dérivée par rapport à $t$.

```python
# Définissons la variable indépendante t et la fonction inconnue y(t)
t = sp.Symbol('t')
y_func = sp.Function('y')(t)

# Définissons l'équation différentielle : y'(t) + y(t) = exp(t)
# On l'écrit sous la forme : y'(t) + y(t) - exp(t) = 0
eq_diff = sp.Eq(y_func.diff(t) + y_func, sp.exp(t))
print("Équation différentielle à résoudre :")
display(eq_diff)

# Résolution générale
solution_generale = sp.dsolve(eq_diff, y_func)
print("\nSolution générale de l'équation :")
display(solution_generale)

# Résolution avec condition initiale, par exemple y(0) = 2
solution_particuliere = sp.dsolve(eq_diff, y_func, ics={y_func.subs(t, 0): 2})
print("\nSolution particulière avec y(0) = 2 :")
display(solution_particuliere)
```

### Étape 9 : Simplifications avancées et Hypothèses

SymPy fournit des outils très ciblés pour simplifier les expressions. Parfois, spécifier des hypothèses (comme "x est réel et positif") permet à SymPy de simplifier des expressions qu'il ne toucherait pas autrement.

```python
# 1. Simplification trigonométrique
expr_trigo = sp.sin(x)**2 + sp.cos(x)**2
print("Expression trigonométrique :")
display(expr_trigo)
print("Simplification :")
display(sp.simplify(expr_trigo))

# 2. L'importance des hypothèses (Assumptions)
# Par défaut, SymPy ne simplifie pas sqrt(a^2) car 'a' pourrait être complexe ou négatif
a = sp.Symbol('a')
print("\nSans hypothèse, sqrt(a^2) =")
display(sp.sqrt(a**2))

# Si on précise que 'p' est réel et positif :
p = sp.Symbol('p', positive=True)
print("Avec l'hypothèse que p est positif, sqrt(p^2) =")
display(sp.sqrt(p**2))
```

### Étape 10 : Systèmes d'équations non-linéaires

Pour résoudre un système d'équations de manière robuste, nous pouvons utiliser `sp.solve` ou `sp.nonlinsolve` en lui transmettant une liste d'équations et la liste des variables à chercher.

```python
# Définissons un système d'équations :
# 1) x^2 + y^2 = 1 (un cercle)
# 2) y = x^2 (une parabole)
eq1 = sp.Eq(x**2 + y**2, 1)
eq2 = sp.Eq(y, x**2)

print("Système d'équations :")
display(eq1, eq2)

# Résolution du système pour x et y
solutions_systeme = sp.solve([eq1, eq2], (x, y))
print("\nSolutions du système (points d'intersection) :")
display(solutions_systeme)
```
